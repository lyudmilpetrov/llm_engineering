````python
# constants

MODEL_GPT = "gpt-4o-mini"
MODEL_LLAMA = "llama3.2"
HEADERS = {"Content-Type": "application/json"}
OLLAMA_API = "http://localhost:11434/api/chat"
````

````python
# set up environment
load_dotenv(override=True)

api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")

openai = OpenAI()
    
````

````python
messages = [
    {"role": "system", "content": "Your are Lead Programmer and your goal is to answer a question from programming. You should respond in a concise and clear manner, providing a brief explanation followed by a simple code example that illustrates the concept."},
    {"role": "user", "content": "REPLACE THIS WITH YOUR QUESTION"}
]

payload = {
        "model": MODEL_GPT,
        "messages": messages,
        "stream": True
    }

question = "What is a function in Python?"
# Function that accepts a question and returns the modified messages list
def set_question(question):
    messages[1]["content"] = question
    return messages
print("You can now use the set_question function to set your question. For example:")
test = set_question(question)
print(test)
# Function that accepts the question and returns the payload for the API request as a dictionary
def get_payload(question, model=MODEL_GPT):
    messages = set_question(question)
    payload = {
        "model": model,
        "messages": messages,
        "stream": True
    }
    return payload
# Test it
payload = get_payload(question, model=MODEL_GPT)
print("Payload for the API request:")
print(json.dumps(payload, indent=2))
````

````python
# Get gpt-4o-mini to answer, with streaming

def get_gpt_answer(payload) -> None:
    stream = openai.chat.completions.create(
    model=payload['model'],
    messages=payload['messages'],
    stream=payload['stream']
    )    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        response = response.replace("```","").replace("markdown", "")
        update_display(Markdown(response), display_id=display_handle.display_id)
    return answer        # type: ignore
# Get the answer
````

````python
answer = get_gpt_answer(payload)
print("Answer from GPT-4o-mini:")
print(answer)
````

````python
# Get Llama 3.2 to answer
payloadForHTTPCall = {
        "model": MODEL_LLAMA,
        "messages": messages,
        "stream": False
    }
response = requests.post(OLLAMA_API, json=payloadForHTTPCall, headers=HEADERS)
# Print output
print(response.json()['message']['content'])
````
